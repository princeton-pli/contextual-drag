### GSM8k test set

```python
import datasets

ds_name = "openai/gsm8k"
ds = datasets.load_dataset(ds_name, 'main', split="test")
print(ds)

ds_new = []
for item in ds:
    problem = item['question']
    answer_raw = item['answer']
    ds_new.append({
        'problem': problem,
        'answer': answer_raw.split('#')[-1].strip(),
        'answer_raw': answer_raw
    })

ds_hf = datasets.Dataset.from_list(ds_new)
ds_hf.save_to_disk('gsm8k/gsm8k.ds')
```
